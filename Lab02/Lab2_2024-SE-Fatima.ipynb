### Task 3 · Randomness and reproducibility
- (a) Set `np.random.seed(2026)`, then generate one random number between 0 and 1.
- (b) Generate 25 numbers from the standard normal distribution.
- (c) Set `np.random.seed(42)` and roll 10,000 Ludo dice with `np.random.randint`. Count each face with `np.unique(..., return_counts=True)`.
- (d) Estimate P(six) and compare it with 1/6.
- (e) Count how many times two sixes came in a row (no loops).

```python
# Task 3 · Randomness and reproducibility

# Task 3a:  
np.random.seed(2026)
print("T3a:", np.random.rand())

# Task 3b: 
print("T3b:", np.random.randn(25))

# Task 3c:
np.random.seed(42)
rolls = np.random.randint(1, 7, size=10000)
faces, counts = np.unique(rolls, return_counts=True)
print("T3c faces:", faces)
print("T3c counts:", counts)

# Task 3d: 
p_six = np.mean(rolls == 6)
print("T3d P(six):", p_six, "| 1/6 =", 1/6, "| difference =", abs(p_six - 1/6))

# Task 3e:
double_six = np.sum((rolls[:-1] == 6) & (rolls[1:] == 6))
print("T3e double sixes:", double_six)
```

### Task 9 · Grace marks
Using `marks`:
- (a) Add 5 grace marks but cap at 100 (`np.clip`).
- (b) Pass % (≥ 50) after grace.
- (c) Index of students who passed **only because of** the grace marks.
- (d) z-score of each student after grace.

```python
# Task 9 · Grace marks

# Task 9a: 
graced = np.clip(marks + 5, 0, 100)
print("T9a:", graced)

# Task 9b:
pass_pct = np.mean(graced >= 50) * 100
print("T9b:", pass_pct, "%")

# Task 9c:
only_grace = np.where((marks < 50) & (graced >= 50))[0]
print("T9c:", only_grace)

# Task 9d: 
z = (graced - graced.mean()) / graced.std()
print("T9d:", np.round(z, 2))
```

### Task 10 · An image is an array

*Optional: `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`*

```python
# Task 10 · An image is an array

# Task 10a:
img = np.zeros((8, 8), dtype=int)
img[1, 1:7] = 255      
img[2, 5] = 255         
img[3, 4] = 255
img[4, 3] = 255
img[5, 3] = 255
img[6, 2] = 255
img[2, 6] = 128        
img[6, 3] = 128
print("T10a:\n", img)

# Task 10b:  
flipped = np.fliplr(img)
print("T10b:\n", flipped)

# Task 10c: 
inverted = 255 - img
print("T10c:\n", inverted)

# Task 10d:  Crop rows 1–4 and columns 3–6.
crop = img[1:5, 3:7]
print("T10d:\n", crop)

# Task 10e: 
bright = np.clip(img + 80, 0, 255)
print("T10e:\n", bright)

# Task 10f: 
print("T10f shape:", img.shape)
print("T10f dtype:", img.dtype)
print("T10f white pixels:", np.sum(img == 255))

# Optional: import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')
import matplotlib.pyplot as plt
plt.imshow(img, cmap='gray')
plt.show()
```
